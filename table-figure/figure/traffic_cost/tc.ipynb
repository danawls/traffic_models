```python
import pandas as pd
import matplotlib.pylab as plt
import numpy as np
```

```python
df = pd.read_csv('yearly_data.csv')
df
```

```python
import matplotlib.font_manager as fm

# 한글 폰트 설정
font_path = '/System/Library/Fonts/AppleSDGothicNeo.ttc'  # 폰트 파일 경로
fontprop = fm.FontProperties(fname=font_path)

# 샘플 데이터 생성
x = df['Year']
y1 = df['Traffic_cost']

# 그래프 스타일 설정
plt.figure(figsize=(8, 5))
plt.plot(x, y1, linewidth=2, linestyle='-', color='#1f77b4')

# 축 라벨 및 타이틀 설정
plt.xlabel('단위: 년', fontsize=10, fontproperties=fontprop)
plt.ylabel('단위: 조원', fontsize=10, fontproperties=fontprop)
plt.title('연도별 교통혼잡비용', fontsize=30, fontweight='bold', fontproperties=fontprop)

# 축 범위 설정
# plt.xlim([0, 10])
# plt.ylim([-1.5, 1.5])

# 격자 설정
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# 범례 설정
plt.legend(fontsize=10, loc='upper right', frameon=False, prop=fontprop)

# 그림 여백 조정
plt.tight_layout()

# 그래프 저장 (고해상도)
plt.savefig('graph.png', dpi=300)

# 그래프 보여주기
plt.show()
```
